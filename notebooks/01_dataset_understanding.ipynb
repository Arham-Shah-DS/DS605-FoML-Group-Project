# Dataset Understanding

This notebook explores the crop disease dataset to understand its
structure, features, target variable, and basic data quality.

The goal is to identify important characteristics of the data
before preprocessing and model training.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate the dataset
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

data_dir = project_root / "data" / "raw"
csv_files = list(data_dir.glob("*.csv"))

assert len(csv_files) == 1, "Expected exactly one CSV file."

df = pd.read_csv(csv_files[0])

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nTarget distribution:")
print(df["Disease_Severity"].value_counts().sort_index())
```

### Disease Distribution

```python
print("Number of disease categories:", df["Disease"].nunique())

display(df["Disease"].value_counts())
```

### Severity distribution and relationship

```python
print("Severity levels:")
display(df["Disease_Severity"].value_counts().sort_index())

print("Disease vs Severity:")
display(pd.crosstab(df["Disease"], df["Disease_Severity"]))
```
